## 2. Dataset Understanding

The dataset contains 4,000 agricultural farm records and 28 columns.

The data consists of 6 categorical variables and 22 numerical variables. The categorical variables include farm location, crop, season, and irrigation method, while the numerical variables describe environmental conditions, soil characteristics, agricultural inputs, production, economic performance, water usage, and disease/pest risk.

The dataset contains:
- 8 states
- 10 districts
- 8 crops
- 3 seasons: Kharif, Rabi, and Zaid
- 4 irrigation methods

The seasonal distribution is uneven, with Kharif having the highest number of records, followed by Rabi and Zaid. Crop records are also distributed unevenly, with Rice having the highest number of observations and Sugarcane the lowest.

The main agricultural performance variables identified for further analysis are Yield_Tonnes_Ha, Production_Tonnes, Revenue_INR, and Profit_INR. Environmental, soil, input, irrigation, and risk variables can be analyzed to understand factors associated with agricultural performance.

```python
# Check missing values and their percentage

missing_summary = pd.DataFrame({
    "Missing_Values": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_summary = missing_summary[missing_summary["Missing_Values"] > 0]
missing_summary
```

```python
missing_rows = df[
    df[["Rainfall_mm", "Soil_Moisture_pct", "Yield_Tonnes_Ha"]]
    .isnull()
    .any(axis=1)
]

print("Number of rows with missing values:", len(missing_rows))

missing_rows.head(20)
```

```python
# Fill missing numerical values with the median

df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(
    df["Soil_Moisture_pct"].median()
)

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())
```

```python
# Verify that the missing values are gone

df[["Rainfall_mm", "Soil_Moisture_pct", "Yield_Tonnes_Ha"]].isnull().sum()
```

```python
# Average agricultural performance for each crop

crop_analysis = df.groupby("Crop")[[
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Profit_INR"
]].mean().sort_values("Profit_INR", ascending=False)

crop_analysis
```

```python
import matplotlib.pyplot as plt #visulaizse

crop_analysis["Profit_INR"].plot(kind="bar", figsize=(10, 5))

plt.title("Average Profit by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average performance by farming season

season_analysis = df.groupby("Season")[[
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Profit_INR"
]].mean().sort_values("Profit_INR", ascending=False)

season_analysis
```

```python
#visulize season by profit
import matplotlib.pyplot as plt
season_analysis["Profit_INR"].plot(kind="bar",figsize=(10,5))
plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average agricultural performance for each district

district_analysis = df.groupby("District")[[
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Profit_INR"
]].mean().sort_values("Profit_INR", ascending=False)

district_analysis
```

#### Recommendations
- Focus on high-performing crops such as Sugarcane where conditions are suitable.
- Investigate the factors contributing to Punjab's higher average yield.
- Improve agricultural practices in lower-yielding regions.
- Improve water-management practices for crops with lower water efficiency.
- Give greater attention to disease and pest management for high-risk crops such as Wheat.
- Use profitability, yield, water efficiency, and risk together when selecting crops.
